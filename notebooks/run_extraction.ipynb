### Add newly loaded data to consolidated data: 

```python


updated_combined, pub_info_table, new_id = add_publication(
    combined=df_cons_corr, 
    new_data=long_df, 
    pub_info_table=pub_info_table,
    publication='Haddock et al. 1999', 
    sample_information=('"Seventy-one patients were recruited to the study, '
                        '48 males (68%), and 23 females (32%)... Fifty-two patients (73%) '
                        'had a diagnosis of schizophrenia and 19 had a diagnosis '
                        'of schizoaffective disorder (27%)." Note on subsample sizes: '
                        '"Patients were also assessed using the PSYRATS (the AH or DS being '
                        'administered where symptoms were present)" (p.882), and '
        '"Forty-two patients had both auditory hallucinations and delusions, 14 had '
                        'hallucinations only and 15 had delusions only" (p.882). '
                        'This gives AHS n = 42+14 = 56 and DS n = 42+15 = 57 (42+14+15 = 71, '
                        'matching the full recruited sample — no patient had neither symptom). '
                        'The 42 who had both symptoms are counted in both subsample sizes, '
                        'so this publication is in ambiguous_publications, '
                        'same as Favrod et al. 2012 and Woodward et al. 2014.'),
    mental_illness_subsamples=[
    'whole_sample_AHS',
    'whole_sample_DS'
], 
    schizophrenia_subsamples=[
    'whole_sample_AHS',
    'whole_sample_DS'
])
                        
```
